 Magic Commands in Jupyter

Magic commands let you time code, run shell commands, install packages, and control the notebook environment.


```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Timing small operations with %timeit

%timeit runs the code many times and reports an average.
Use it for small, fast operations.


```python
%timeit np.random.randn(10000)
```

## 2. Timing longer operations with %%time

%%time runs the entire cell once.
Use it for training loops or expensive computations.


```python
%%time
# Simulate a "training loop"
total = 0
for i in range(5_000_000):
    total += i
total
```

## 3. Enabling inline plots

%matplotlib inline makes every plt.plot() show directly in the notebook.


```python
%matplotlib inline

x = np.linspace(0, 10, 200)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.show()
```

## 4. Running shell commands with !

The ! prefix lets you run terminal commands directly from the notebook.
Here we install a package.


```python
!uv pip install scikit-learn

```
