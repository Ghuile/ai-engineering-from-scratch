## Why DataFrames look so good in notebooks

Pandas integrates with Jupyter to render DataFrames as HTML tables.
This gives you sortable, readable, visually clean tables without extra code.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot([1, 2, 3, 4], [1, 4, 2, 3], marker='o')
plt.title("Inline Plot Example")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid(True)
plt.show()
```

## Inline Plots

Matplotlib integrates with Jupyter so that every plot appears directly in the notebook.
This is essential for visualizing model performance, data distributions, and results.


```python
from IPython.display import Image, display

# Replace with your own image file
display(Image(filename="Smiley.png"))
```
